```python
import matplotlib.pyplot as plt
import numpy as np

class GraficoPoblacionRusia:
    """
   La población total de Rusia ronda los 146 millones de habitantes, concentrándose principalmente en la parte occidental del país. La Zona Norte, que incluye regiones como Murmansk, Arkhangelsk, Karelia y Komi, forma parte del Distrito Federal Noroeste y reúne aproximadamente 13.9 millones de personas, lo que la convierte en un área extensa pero de muy baja densidad poblacional, dominada por climas fríos y ciudades aisladas.

La Zona Centro, correspondiente al Distrito Federal Central, es la más poblada del país con cerca de 40.1 millones de habitantes. Aquí se ubican Moscú y otras grandes ciudades que funcionan como núcleo político, económico y cultural de Rusia. Esta zona destaca por su alta urbanización, mayor oferta laboral y mejor infraestructura, lo que explica su fuerte concentración demográfica.

La Zona Sur, entendida como la combinación entre el Distrito Federal Sur y el Distrito del Cáucaso Norte, suma alrededor de 26.9 millones de habitantes. Se caracteriza por un clima más templado, mayor actividad agrícola y una densidad poblacional media, apoyada en regiones como Krasnodar, Rostov del Don, Volgogrado y las repúblicas caucásicas. Aunque menos poblada que el Centro, es una zona dinámica y con un crecimiento demográfico más estable

    """

    def __init__(self):
        self.poblacion_por_zona = {
            'Zona Norte': 7.2,
            'Zona Centro': 47.5,
            'Zona Sur': 24.3
        }

    def crear_grafico(self):
        fig, ax = plt.subplots()
        zonas = list(self.poblacion_por_zona.keys())
        poblaciones = list(self.poblacion_por_zona.values())

        ax.pie(
            poblaciones,
            labels=zonas,
            autopct='%1.1f%%',
            startangle=90
        )

        ax.set_title('Distribución de población por zonas (Rusia)')
        ax.axis('equal')
        plt.show()


grafico = GraficoPoblacionRusia()
grafico.crear_grafico()
```
