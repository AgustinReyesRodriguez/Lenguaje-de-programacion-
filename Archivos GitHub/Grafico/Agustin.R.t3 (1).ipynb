```python
import matplotlib.pyplot as plt


zonas = ["Norte", "Centro", "Sur"]
poblacion = [7.2, 47.5, 24.3]  #  millones


plt.style.use("seaborn-v0_8-whitegrid")

plt.figure(figsize=(8, 5))
barras = plt.bar(
    zonas, poblacion,
    color=["#6baed6", "#74c476", "#fd8d3c"],  
    edgecolor="black", linewidth=1.2
)


plt.title("Población de Rusia por Zona Geográfica", fontsize=19, fontweight='bold', color="#333")
plt.xlabel("Zonas de Rusia", fontsize=12)
plt.ylabel("Población (millones de habitantes)", fontsize=12)


for i, bar in enumerate(barras):
    plt.text(
        bar.get_x() + bar.get_width()/2,    
        bar.get_height() + 0.8,             
        f"{poblacion[i]} M",
        ha="center", fontsize=11, fontweight='bold', color="#222"
    )




plt.tight_layout()
plt.show()
"""
Zona Norte
Incluye regiones como Murmansk

, Arkhangelsk, Karelia y Komi.
Es un área muy fría y con poca población repartida en grandes extensiones de tierra.
La población total se estima en unos 7.2 millones de habitantes.

Zona Centro
Aquí se concentra la mayor parte de la población con ciudades como Moscú, an Petersburgo, Tver, olensk y Nizhni Nóvgorod.
Es el corazón político, económico y cultural del país donde se desarrollan la mayoría de las actividades.
Su población ronda los 47.5 millones de personas.

Zona Sur
Incluye regiones como Krasnodar, Rostov del Don, Volgogrado y el Cáucaso Norte.
Tiene un clima más templado con mucha actividad agrícola y varias ciudades importantes.
La población se calcula en aproximadamente 24.3 millones de habitantes.
"""
```
